# Round 2: Data Visualization

**Goal**: Visualise the combined dataset including Primary and Secondary emotions.
- Distribution of Primary Emotions
- Distribution of Secondary Emotions
- Co-occurrence Heatmap (Primary vs Secondary)
- Language-wise Samples

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style='whitegrid', palette='muted')
print("Libraries imported.")
```

## 1. Load Data

```python
df = pd.read_excel('/Users/hemishjain22/Desktop/Hackathons/hack4healtj/round2/Combined_Emotions.xlsx')
print(f"Total Samples: {len(df)}")
display(df.head())
```

## 2. Primary Emotion Distribution

```python
plt.figure(figsize=(15, 8))
primary_counts = df['Primary'].value_counts()
sns.barplot(x=primary_counts.index, y=primary_counts.values, palette='viridis')
plt.title('Distribution of Primary Emotions', fontsize=16)
plt.xticks(rotation=90)
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## 3. Secondary Emotion Distribution

```python
plt.figure(figsize=(15, 8))
secondary_counts = df['Secondary'].value_counts()
sns.barplot(x=secondary_counts.index, y=secondary_counts.values, palette='magma')
plt.title('Distribution of Secondary Emotions', fontsize=16)
plt.xticks(rotation=90)
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## 4. Primary vs Secondary Co-occurrence

```python
# Create a crosstab
crosstab = pd.crosstab(df['Primary'], df['Secondary'])

plt.figure(figsize=(20, 15))
sns.heatmap(crosstab, cmap='Reds', linewidths=0.5)
plt.title('Co-occurrence of Primary and Secondary Emotions', fontsize=20)
plt.xlabel('Secondary Emotion')
plt.ylabel('Primary Emotion')
plt.tight_layout()
plt.show()
```

## 5. Sample Poems by Language

```python
unique_langs = df['Source'].unique()
for lang in unique_langs:
    print(f"\n{'='*20} {lang} {'='*20}")
    display(df[df['Source'] == lang].head(5))
```
